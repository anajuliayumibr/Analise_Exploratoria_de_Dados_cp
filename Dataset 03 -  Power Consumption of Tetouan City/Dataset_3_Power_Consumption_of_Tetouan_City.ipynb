###Dataset 3 — Power Consumption of Tetouan City (Ana ju)

```python
import pandas as pd
```

Fonte: https://archive.ics.uci.edu/dataset/849/power+consumption+of+tetouan+city

Situação

Uma equipe responsável pelo planejamento da distribuição de energia em Tétouan deseja identificar qual das três zonas apresenta o maior pico de consumo e observar as condições ambientais registradas nos momentos de maior demanda.

```python
dados = pd.read_csv('OFTetuan_City_power_consumption_-_amostra_15__CERTO.csv')
```

1. Renomeie as três variáveis de consumo para Consumo_Zona_1, Consumo_Zona_2 e
Consumo_Zona_3.

```python
dados = dados.rename({'Zone 1 Power Consumption': 'Consumo_Zona_1',
                      'Zone 2  Power Consumption': 'Consumo_Zona_2',
                      'Zone 3  Power Consumption': 'Consumo_Zona_3'},
                     axis=1)

# Convert the power consumption columns to numeric, coercing errors
power_cols = ['Consumo_Zona_1', 'Consumo_Zona_2', 'Consumo_Zona_3']
for col in power_cols:
    dados[col] = pd.to_numeric(dados[col], errors='coerce')
```

```python
dados.columns
```

2. Determine o consumo máximo registrado em cada uma das três zonas.

```python
max_consumo = dados[['Consumo_Zona_1', 'Consumo_Zona_2', 'Consumo_Zona_3']].max()
max_consumo
```

3. Identifique qual zona apresenta o maior pico de consumo na amostra.

```python
zona_maior_pico = max_consumo.idxmax()
print(f"A zona com o maior pico de consumo é: {zona_maior_pico}")
```

  4. Para a zona identificada, calcule 70% do valor máximo e crie um DataFrame com os registros acima desse limiar.

```python
max_consumo_zona_maior_pico = max_consumo[zona_maior_pico]
limiar_70_pico = 0.70 * max_consumo_zona_maior_pico

df_alta_consumo = dados[dados[zona_maior_pico] > limiar_70_pico].copy()

print(f"Valor máximo de consumo para {zona_maior_pico}: {max_consumo_zona_maior_pico:.2f}")
print(f"70% do valor máximo de consumo para {zona_maior_pico}: {limiar_70_pico:.2f}")
print(f"DataFrame com registros acima do limiar de 70% (df_alta_consumo):")
display(df_alta_consumo.head())
print(f"Número de registros em df_alta_consumo: {len(df_alta_consumo)}")
```

5. Conte os registros e calcule o percentual que eles representam.

```python
num_registros_alta_consumo = len(df_alta_consumo)
total_registros_originais = len(dados)

percentual_alta_consumo = (num_registros_alta_consumo / total_registros_originais) * 100

print(f"Número de registros de alta consumo: {num_registros_alta_consumo}")
print(f"Total de registros no dataset original: {total_registros_originais}")
print(f"Percentual de registros de alta consumo: {percentual_alta_consumo:.2f}%")
```

6. Calcule a temperatura média e crie um segundo DataFrame contendo consumo acima do limiar e temperatura acima da média.

```python
media_temperatura = dados['Temperature'].mean()

df_alta_consumo_temp_alta = df_alta_consumo[df_alta_consumo['Temperature'] > media_temperatura].copy()

print(f"Temperatura média no dataset original: {media_temperatura:.2f}°C")
print(f"DataFrame com registros de alta consumo e temperatura acima da média (df_alta_consumo_temp_alta):")
display(df_alta_consumo_temp_alta.head())
print(f"Número de registros em df_alta_consumo_temp_alta: {len(df_alta_consumo_temp_alta)}")
```

7. Compare os dois conjuntos e explique o que ocorreu com a quantidade de registros após a inclusão da condição ambiental.

```python
num_registros_alta_consumo_temp_alta = len(df_alta_consumo_temp_alta)

percentual_alta_consumo_temp_alta = (num_registros_alta_consumo_temp_alta / total_registros_originais) * 100

print(f"Número de registros em df_alta_consumo (apenas alta consumo): {num_registros_alta_consumo}")
print(f"Percentual de registros em df_alta_consumo: {percentual_alta_consumo:.2f}%")
print("\n")
print(f"Número de registros em df_alta_consumo_temp_alta (alta consumo E alta temperatura): {num_registros_alta_consumo_temp_alta}")
print(f"Percentual de registros em df_alta_consumo_temp_alta: {percentual_alta_consumo_temp_alta:.2f}%")

reducao_registros = num_registros_alta_consumo - num_registros_alta_consumo_temp_alta
percentual_reducao = (reducao_registros / num_registros_alta_consumo) * 100

print("\n")
print(f"Ao incluir a condição de temperatura acima da média, o número de registros reduziu em: {reducao_registros} ({-percentual_reducao:.2f}%) registros.")
print(f"Isso significa que {percentual_reducao:.2f}% dos registros de alta consumo ocorrem com temperatura igual ou abaixo da média.")
```
