Situação

Uma equipe de operação de uma usina fotovoltaica deseja localizar períodos de alta geração e identificar quais inversores aparecem com maior frequência nesses momentos. Nesta atividade, o foco inicial será o arquivo de geração da planta, sem realizar junção com o arquivo de sensores.

```python
df4 = pd.read_csv('/content/Plant_1_Generation_Data_Tratada.csv')
```

```python
#1.	Carregue a amostra e renomeie as principais variáveis para nomes mais claros, como Potencia_CC, Potencia_CA e Geracao_Diaria.
df4.rename(columns={'DATE_TIME': 'Hora_Dia',
                    'PLANT_ID': 'Planta_ID',
                    'SOURCE_KEY': 'Chave_Origem',
                    'AC_POWER': 'Potencia_CC',
                    'DC_POWER': 'Potencia_CA',
                    'DAILY_YIELD': 'Geracao_Diaria',
                    'TOTAL_YIELD': 'Geracao_Total',
                    }, inplace=True)
```

```python
#2.	Apresente head(), shape, info() e describe().
df4.head(), df4.shape, df4.info(), df4.describe()
```

```python
#3. Determine a maior potência CA registrada.
df4=['Potencia_CA'].max()
```

```python
#4. Calcule um limiar correspondente a 70% da potência CA máxima.
PMAX = df4['Potencia_CA'].max()
P70 = 0.7 * PMAX
limiar_max = df4[df4['Potencia_CA'] > P70]
limiar_max
```

```python
#5. Crie um DataFrame contendo somente registros acima desse limite e calcule sua quantidade e percentual.
# Novo DataFrame!
df4novo = df4[df4['Potencia_CA'] > P70]
df4novo.head()

# Quantidade de registros selecionados (consumo acima de 70% do máximo)
qtd_selecionados = len(df4novo)
qtd_total = len(df4)

df4_percentual = (qtd_selecionados / qtd_total) * 100

print(f'Registros selecionados: {qtd_selecionados} de {qtd_total}.')
print(f'O percentual da amostra representa {df4_percentual:.2f}%.')
```

```python
#6 Utilize value_counts() em SOURCE_KEY dentro do DataFrame de alta geração.
contagem_inversores = df4novo["Chave_Origem"].value_counts()

print(contagem_inversores)
```

```python
#7 Identifique qual inversor aparece com maior frequência nos registros de alta geração e descreva o que esse resultado permite observar, sem concluir desempenho ou falha apenas com esse recorte.
inversor_mais_frequente = contagem_inversores.idxmax()
maior_frequencia = contagem_inversores.max()

print("Inversor mais frequente:", inversor_mais_frequente)
print("Quantidade de registros:", maior_frequencia)
```
