Situação

A equipe de gestão de energia de uma indústria siderúrgica deseja localizar situações de consumo elevado e verificar se esses períodos coincidem com condições de carga elevada ou valores menos favoráveis de fator de potência.

```python
df2 = pd.read_csv('/content/EX2_Steel_industry_data.csv')
```

##Carregue a amostra e renomeie Usage_kWh para Consumo_kWh, simplificando também os nomes dos atributos de fator de potência.

```python
df2.rename(columns={'Usage_kWh': 'Consumo_kWh',
            'Lagging_Current_Power_Factor': 'Fator_corrente_atrasada',
            'Leading_Current_Power_Factor': 'Fator_Potencia',
            'Lagging_Current_Reactive.Power_kVarh': 'Potencia_Reativa_atrasada',
            'Leading_Current_Reactive_Power_kVarh': 'Potencia_Reativa',
            'WeekStatus': 'Status_Reativa',
            'Day_of_week': 'Dia_Semana'
            }, inplace=True)
```

##Determine o maior consumo registrado e calcule um limiar de 75% do consumo máximo.

```python
df2['Consumo_kWh'].max()
```

```python
PMAX = df2['Consumo_kWh'].max()
P75 = 0.75 * PMAX
limiar_max = df2[df2['Consumo_kWh'] > P75]
limiar_max
```

```python
limiar_max.shape
```

##Crie um DataFrame contendo os registros acima desse limiar e calcule a quantidade e o percentual de registros.

```python
df2_1 = df2[df2['Consumo_kWh'] > P75]
df2.head()
```

```python
qtd_selecionados = len(df2_1)
qtd_total = len(df2)

df2_1_percentual = (qtd_selecionados / qtd_total) * 100

print(f'Registros selecionados: {qtd_selecionados} de {qtd_total}.')
print(f'O percentual da amostra representa {df2_1_percentual:.2f}%.')
```

##Verifique quantos registros desse DataFrame pertencem à categoria Maximum Load.

```python
qtd_maximum_load = df2_1[df2_1['Load_Type'] == 'Maximum_Load'].shape[0]

print(f'Registros da categoria Maximum_Load: {qtd_maximum_load} de {len(df2_1)} registros de consumo elevado.')
```

##Escolha um dos atributos de fator de potência, observe seus valores e defina um limite coerente para representar valores mais baixos.

```python
df2['Fator_Potencia'].describe()
limite_fp = 50
```

##Crie um novo DataFrame contendo simultaneamente consumo elevado e fator de potência abaixo do limite escolhido.

```python
df2_2 = df2[(df2['Consumo_kWh'] > P75) & (df2['Fator_Potencia'] < limite_fp)]

qtd_selecionados = len(df2_2)
qtd_total = len(df2)
df2_2_percentual = (qtd_selecionados / qtd_total) * 100

print(f'Registros selecionados: {qtd_selecionados} de {qtd_total}.')
print(f'O percentual da amostra representa {df2_2_percentual:.2f}%.')
```

##Explique por que esse segundo conjunto de registros pode merecer maior atenção da equipe de energia.

Esse cruzamento (consumo alto + fator de potência baixo) seria o cenário mais crítico pra equipe de energia, pois indica ineficiência no uso da energia e pode gerar multas na conta. Porém, nesta amostra o resultado deu 0 registros: quando o consumo está no pico, o Fator_Potencia (Leading) fica sempre saturado em 100, ou seja, esse atributo específico não é o responsável por ineficiência nos momentos de consumo elevado.
